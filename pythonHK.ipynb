```python
# 사용자로부터 입력받기
num1 = float(input("첫 번째 숫자를 입력하세요: "))
operator = input("연산자를 입력하세요 (+, -, *, /): ")
num2 = float(input("두 번째 숫자를 입력하세요: "))

# 연산자에 따른 결과 계산
if operator == '+':
    result = num1 + num2
elif operator == '-':
    result = num1 - num2
elif operator == '*':
    result = num1 * num2
elif operator == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: 0으로 나눌 수 없습니다."
else:
    result = "Error: 올바르지 않은 연산자입니다."

# 결과 출력
print(f"결과: {result}")
```

```python
import yfinance as yf

# 1. 설정 변수
ticker_symbol = "NVDA"
start_date = "2020-01-02"  # 2020년 첫 거래일
initial_investment_krw = 10_000_000  # 원화 1,000만 원

print("야후 파이낸스에서 데이터를 불러오는 중...")

# 2. NVDA 주가 데이터 다운로드
ticker = yf.Ticker(ticker_symbol)
df = ticker.history(start=start_date)

# 3. 매수 시점(2020년 첫 거래일) 및 현재 시점 종가 가져오기
initial_price_usd = df.iloc[0]['Close']
current_price_usd = df.iloc[-1]['Close']
current_date = df.index[-1].strftime('%Y-%m-%d')

# 4. 수익률 계산 (주식 분할 등은 yfinance의 history 데이터에 자동 반영됨)
return_rate = (current_price_usd - initial_price_usd) / initial_price_usd
current_investment_krw = initial_investment_krw * (1 + return_rate)
profit_krw = current_investment_krw - initial_investment_krw

# 5. 결과 출력
print("=" * 50)
print(f"  [엔비디아(NVDA) 2020년 투자 수익률 계산]")
print("=" * 50)
print(f"• 매수일 기준 주가 ({start_date[:10]}): ${initial_price_usd:.2f}")
print(f"• 최근 기준 주가   ({current_date}): ${current_price_usd:.2f}")
print(f"• 주가 상승률: {return_rate * 100:,.2f}%")
print("-" * 50)
print(f"• 원금: {initial_investment_krw:,.0f} 원")
print(f"• 현재 평가 금액: {current_investment_krw:,.0f} 원")
print(f"• 총 수익: {profit_krw:,.0f} 원")
print("=" * 50)
```
